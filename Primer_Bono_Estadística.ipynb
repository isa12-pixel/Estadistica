# Manejo de módulos integrados

NumPy: Se usa principalmente para trabajar con datos numéricos de forma rápida, eficiente y estructurada. Un objeto tipo numpy es un arreglo base para hacer matemáticas y manipular en Python.

Pandas se usa para trabajar con datos estructurados (Como tablas de Excel o bases de datos). Un objeto tipo para cargar, limpiar, transformar y analizar datos.

La diferencia entre NumPy y Pandas radica en su propósito y nivel de uso: NumPy es una librería enfocada en el cálculo numérico eficiente mediante arrays multidimensionales, ideal para operaciones matemáticas rápidas, álgebra lineal y procesamiento de datos puramente numéricos; mientras que Pandas es una librería de más alto nivel diseñada para la manipulación y análisis de datos estructurados como tablas tipo Excel o bases de datos, permitiendo trabajar con columnas con nombres, datos de distintos tipos, limpieza, filtrado y agrupación; en la práctica, Pandas se apoya internamente en NumPy, por lo que ambos se complementan y se utilizan juntos en la mayoría de flujos de ciencia de datos.

```python
dfn=pd.DataFrame({
    'id' : [0,1,2,3,4],
    'nombre' : ['Jimin','Tae','RM','Jin',"Yoongi"],
    'ciudad' : ['Busan','Seúl','Daegu','Jeonju','Ulsan'],
    'edad' : [33, 30,30,31,33],
    'salario' : [50000,30000,45000,45000,50000]
})
dfn
```

```python
dfn['salario_anual']=dfn['salario']*12
#aplicar funcion
dfn['edad_doble']=dfn['edad'].apply(lambda x:x*2)
dfn
```

```python
dfn=dfn.rename(columns={'edad':'age'})
dfn
```

```python
dfn['nombre']=dfn['nombre'].str.upper()
dfn['nombre']=dfn['nombre'].str.strip()
dfn['nombre']=dfn['nombre'].str.replace('GABRIELA','GABRIEL')
dfn
```

```python
grupo=dfn.groupby('ciudad')['salario'].mean() # Toma los elemento de la ciudad, y se calcula la media aritmetica
grupo
```

```python
ordenado=dfn.sort_values(by='age') # ordena por la edad de menor a mayor
ordenado
```
